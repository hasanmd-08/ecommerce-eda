````python
#2. Identifying the top selling category
Top_Performer = df.groupby('Product_Category')['Total_Sales'].sum().sort_values(ascending=False)
print(f'The top selling category is: {Top_Performer.index[0]}, with a total sales of ${Top_Performer.iloc[0]}')
````

````python
#3. Customer Satisfaction Analysis
average_rating = df['Customer_Rating'].mean()
print(f"The average product rating across all products is: {average_rating:.2f}")
````

Based on the average customer rating, we can interpret the overall customer satisfaction:
- An average rating closer to 5 indicates high customer satisfaction.
- An average rating closer to 1 indicates low customer satisfaction.

Further analysis might involve looking at the distribution of ratings or identifying specific products with very low or very high ratings to understand the drivers of satisfaction.

````python
#4. Premium Product Filtering
premium_products = df[df['Customer_Rating'] > 4]
print('Premium Products (Rating > 4):')
premium_products
````

````python
#Importing the library
import matplotlib.pyplot as plt
````

````python
#5. Visualization – Sales Performance
# Aggregate total sales by Product_Category
category_sales = df.groupby('Product_Category')['Total_Sales'].sum()

# Create a bar chart with formatted for better readability
plt.figure(figsize=(10, 6))
plt.bar(category_sales.index, category_sales.values, color='skyblue')
plt.xlabel('Product Category')
plt.ylabel('Total Sales ($)')
plt.title('Total Sales per Product Category')
plt.xticks(rotation=45)
plt.grid(axis='y', linestyle='--', alpha=0.7) # Add a grid for easier comparison
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
````

````python
#6. Pricing Analysis using histogram
plt.hist(df['Price'], bins=10, color='lightgreen', edgecolor='black')
plt.xlabel('Price')
plt.ylabel('Frequency')
plt.title('Distribution of Product Prices')
plt.grid(axis='y', linestyle='--', alpha=0.6)
plt.show()
````
